# Explore manysearch results with different intersect & frequency thresholds

```python
import polars as pl
```

```python
base_df = (
    pl.scan_csv('../outputs.cds3/mag+gtdb.cds3.x.3216.filtered.csv')
).collect()
```

```python
def filter_and_summarize_core(df, *, threshold=5):
    df = df.filter(pl.col('intersect_hashes') >= threshold)

    n_acc = df['match_name'].n_unique()
    by_species = df.group_by('query_name').agg(
        freq=pl.len() / n_acc
    )
    return by_species
    
```

```python
xx = []
for intersect_threshold in range(5, 55, 5):
    dd = dict(intersect_threshold=f"{intersect_threshold}k")
    for freq_threshold in range(80, 105, 5):
        sum_df = filter_and_summarize_core(base_df, threshold=intersect_threshold)
        sum_df = sum_df.filter(pl.col('freq') >= freq_threshold / 100)
        print(intersect_threshold, freq_threshold, len(sum_df))
        dd[f"{freq_threshold}%"] = len(sum_df)
    xx.append(dd)

xx_df = pl.DataFrame(xx)
```
