```python
import polars as pl
```

```python
base_df = (
    pl.scan_parquet('../outputs.cds3/pq/mag+gtdb.cds3.x.3216.manysearch.parquet')
    .filter(pl.col('intersect_hashes') >= 5)
).collect()
```

```python
def filter_and_summarize_core(df, *, threshold=5):
    df = df.filter(pl.col('intersect_hashes') >= threshold)

    n_acc = df['match_name'].n_unique()
    by_species = df.group_by('query_name').agg(
        freq=pl.len() / n_acc
    )
    return by_species
    
```

```python
for threshold in range(5, 55, 5):
    sum_df = filter_and_summarize_core(df, threshold=threshold)
    sum_df = sum_df.filter(pl.col('freq') >= 0.95)
    print(threshold, len(sum_df))
    if threshold >= 35:
        print(sum_df['query_name'])
```
