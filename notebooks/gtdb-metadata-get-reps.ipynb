# Extract the GTDB representative genomes for the 16 core species

Primarily used for the PRJEB88111 nanopore analysis, but maybe more useful down the road?

```python
import polars as pl
import screed
```

## Read in the GTDB metadata for bacteria 

```python
df = pl.read_csv('../bac120_metadata_r226.tsv.gz', separator='\t', ignore_errors=True)
```

## Grab the species names of the core species

```python
CORE_NAMES=set([ x.strip() for x in open('../inputs.branchwater/names.list') ])
```

## Extract info from GTDB metadata table

```python
df2 = df.filter(
    (pl.col('gtdb_taxonomy').str.split(';').list.get(-1).is_in(CORE_NAMES)) &
    (pl.col('gtdb_representative') == 't')
)
```

## Get the accessions and save them to a CSV...

```python
df3 = df2.with_columns(
    acc=pl.col('gtdb_genome_representative').str.slice(3)
)
assert len(df3) == len(CORE_NAMES)
#df3.write_csv('../inputs.PRJEB88111.nanopore/reps.csv')
```

## ...and print them out nicely, too.

```python
print("\n".join(df3['acc']))
```
