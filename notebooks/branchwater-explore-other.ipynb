# Explore the 'other' category of branchwater matches

```python
import polars as pl
```

```python
# produced by branchwater-figure.ipynb
found_df = pl.read_csv('branchwater_8species_other.csv')
```

```python
# get metadata, only compare WGS
metadata_df = (
    pl.scan_parquet("/group/ctbrowngrp5/sra-metagenomes/20241128-metadata.parquet")
    .filter(pl.col("acc") != "NP")
    .filter(pl.col("assay_type") == "WGS")
#    .select(["acc", "organism", "bioproject", "mbases", "host", "project_name"]) 
    .collect()
)
print(f"found metadata records for {len(metadata_df)} WGS metagenome data sets from the SRA")
```

```python
found_df2 = found_df.join(metadata_df, on=['acc', 'bioproject', 'organism'], how='inner')
```

```python
all_bioproject_sum_df = metadata_df.group_by('bioproject').agg(
    n_acc_total=pl.len(),
    orgs=pl.col('organism').unique(),
)
```

```python
found_df2.group_by('bioproject').agg(
    n_acc_matched=pl.len(),
).join(all_bioproject_sum_df, on='bioproject', how='inner').sort('n_acc_matched').filter(pl.col('n_acc_matched') > 2)
```

## Explore PRJNA506496, fermentation metagenome

This study aims to investigate what microbes capable of utilizing plant cell wall polysaccharides and their functions. The submitted metagenomic sequences include three in vitro fermentation experiments (P1, P2 & H). Of P1, wheat arabinoxylan and rye arabinoxylan were fermented for up to 48 h with a porcine faecal inoculum. Of P2, wheat arabinoxylan and tamarind galactoxyloglucan were fermented for up 72 h with a porcine fecal inoculum. For H, wheat arabinoxylan and tamarind galactoxyloglucan were fermented for up 48 h with a human fecal inoculum. The polysaccharides were the only carbon source in the medium for microbial growth.

all 32 pig (P1 and P2) were found; all 7 H not.

```python
found_df2.filter(pl.col('bioproject') == 'PRJNA506496')
```

```python
found_df2.filter(pl.col('bioproject') == 'PRJNA506496').with_columns(
    exp=pl.col('library_name').str.split('_').list.get(0)
).sort('exp')['exp'].unique()
```

```python
metadata_df.filter(pl.col('bioproject') == 'PRJNA506496').with_columns(
    exp=pl.col('library_name').str.split('_').list.get(0)
)['exp'].value_counts()
```
