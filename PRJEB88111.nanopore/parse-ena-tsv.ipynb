## Look at the file report and also check md5sums in nanopore directory

```python
import polars as pl
```

```python
runs_df = pl.read_csv('filereport_read_run_PRJEB88111-md5.tsv', separator='\t')
```

```python
runs_df
```

```python
md5_table = runs_df.select(['fastq_md5', 'run_accession'])
md5_table.head(1)
```

## Write out md5s per ENA file report

```python
with open('ENA-fastq-md5sum.txt', 'wt') as fp:
    for row in md5_table.sort('run_accession').iter_rows(named=True):
        print(f"{row['fastq_md5']}  {row['run_accession']}.fastq.gz", file=fp)
```

## Read in actual md5 sums

calculated via `md5sum *.fastq.gz > fastq-md5sum.txt.new `

```python
md5_xx = []
for row in open('fastq-md5sum.txt'):
    md5, filename = row.strip().split('  ')
    acc = filename.split('.')[0]
    md5_xx.append(dict(actual_md5=md5, run_accession=acc))

actual_md5_df = pl.DataFrame(md5_xx)
actual_md5_df
```

## Join actual with expected and compare

```python
join_df = md5_table.join(actual_md5_df, how='left', on='run_accession').fill_null('xx')
join_df.head(1)
```

```python
bad_acc = join_df.filter(pl.col('fastq_md5') != pl.col('actual_md5'))
bad_acc
```

```python
## print out all the actually bad md5s
```

```python
xx = [ f"{acc}.fastq.gz" for acc in bad_acc['run_accession'] ]

print(' '.join(xx))
```
